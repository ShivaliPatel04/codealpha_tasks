```python
df1 = pd.read_csv("/content/Unemployment in India.csv")
df2 = pd.read_csv("/content/Unemployment_Rate_upto_11_2020.csv")
```

```python
print("===== DATASET 1 =====")
print("Columns:")
print(df1.columns.tolist())

print("\nMissing values:")
print(df1.isnull().sum())

print("\n===== DATASET 2 =====")
print("Columns:")
print(df2.columns.tolist())

print("\nMissing values:")
print(df2.isnull().sum())
```

```python
df1.info()
print("\n--------------------\n")
df2.info()
```

```python
df1.columns = df1.columns.str.strip()
df2.columns = df2.columns.str.strip()

print("Dataset 1 columns:")
print(df1.columns.tolist())

print("\nDataset 2 columns:")
print(df2.columns.tolist())
```

```python
print("Dataset 1 missing values:")
print(df1.isnull().sum())

print("\nDataset 2 missing values:")
print(df2.isnull().sum())
```

```python
df1 = df1.dropna()
df2 = df2.dropna()

print("Dataset 1 shape after cleaning:", df1.shape)
print("Dataset 2 shape after cleaning:", df2.shape)
```

```python
df1['Date'] = pd.to_datetime(df1['Date'], dayfirst=True)
df2['Date'] = pd.to_datetime(df2['Date'], dayfirst=True)

print(df1['Date'].head())
print(df2['Date'].head())
```

```python
print("Dataset 1 date range:")
print(df1['Date'].min(), "to", df1['Date'].max())

print("\nDataset 2 date range:")
print(df2['Date'].min(), "to", df2['Date'].max())
```

```python
df1.describe()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(12,5))

sns.lineplot(
    data=df1,
    x='Date',
    y='Estimated Unemployment Rate (%)'
)

plt.title('Unemployment Rate in India Over Time')
plt.xlabel('Date')
plt.ylabel('Unemployment Rate (%)')
plt.xticks(rotation=45)
plt.show()
```

```python
highest = df1.loc[
    df1['Estimated Unemployment Rate (%)'].idxmax()
]

lowest = df1.loc[
    df1['Estimated Unemployment Rate (%)'].idxmin()
]

print("Highest Unemployment Rate:")
print(highest)

print("\nLowest Unemployment Rate:")
print(lowest)
```

```python
average_unemployment = df1['Estimated Unemployment Rate (%)'].mean()

print("Average Unemployment Rate:",
      round(average_unemployment, 2), "%")
```

```python
region_avg = df1.groupby('Region')['Estimated Unemployment Rate (%)'].mean()

region_avg = region_avg.sort_values(ascending=False)

print(region_avg)
```

```python
plt.figure(figsize=(12,7))

region_avg.plot(kind='bar')

plt.title('Average Unemployment Rate by Region')
plt.xlabel('Region')
plt.ylabel('Average Unemployment Rate (%)')
plt.xticks(rotation=90)
plt.show()
```

```python
print("Highest average unemployment:")
print(region_avg.index[0], ":", round(region_avg.iloc[0], 2), "%")

print("\nLowest average unemployment:")
print(region_avg.index[-1], ":", round(region_avg.iloc[-1], 2), "%")
```

```python
covid_start = pd.Timestamp('2020-03-25')

before_covid = df1[df1['Date'] < covid_start]
during_covid = df1[df1['Date'] >= covid_start]

print("Average unemployment before COVID/lockdown:",
      round(before_covid['Estimated Unemployment Rate (%)'].mean(), 2), "%")

print("Average unemployment during COVID/lockdown:",
      round(during_covid['Estimated Unemployment Rate (%)'].mean(), 2), "%")
```

```python
comparison = pd.Series({
    'Before COVID': before_covid['Estimated Unemployment Rate (%)'].mean(),
    'During COVID': during_covid['Estimated Unemployment Rate (%)'].mean()
})

comparison.plot(kind='bar')

plt.title('Unemployment Rate: Before vs During COVID-19')
plt.xlabel('Period')
plt.ylabel('Average Unemployment Rate (%)')
plt.xticks(rotation=0)
plt.show()
```

```python
monthly_avg = df1.groupby(
    df1['Date'].dt.to_period('M')
)['Estimated Unemployment Rate (%)'].mean()

monthly_avg.index = monthly_avg.index.to_timestamp()

plt.figure(figsize=(12,5))

plt.plot(monthly_avg.index, monthly_avg.values, marker='o')

plt.title('Monthly Average Unemployment Rate in India')
plt.xlabel('Month')
plt.ylabel('Unemployment Rate (%)')
plt.xticks(rotation=45)
plt.grid(True)

plt.show()
```

```python
area_avg = df1.groupby('Area')['Estimated Unemployment Rate (%)'].mean()

print(area_avg)
```

```python
plt.figure(figsize=(7,5))

area_avg.plot(kind='bar')

plt.title('Average Unemployment Rate: Rural vs Urban')
plt.xlabel('Area')
plt.ylabel('Average Unemployment Rate (%)')
plt.xticks(rotation=0)

plt.show()
```

```python
correlation = df1[
    ['Estimated Unemployment Rate (%)',
     'Estimated Employed',
     'Estimated Labour Participation Rate (%)']
].corr()

print(correlation)
```

# Unemployment Analysis with Python

## Project Overview

This project analyzes unemployment data in India using Python. The analysis focuses on unemployment trends over time, regional variations, rural and urban differences, and the impact of the COVID-19 pandemic.

The project includes data cleaning, exploratory data analysis, statistical analysis, and data visualization using Pandas, Matplotlib, and Seaborn.

## Key Findings

- The overall average unemployment rate in the analyzed dataset was 11.79%.
- The average unemployment rate before the COVID-19 lockdown period was 9.51%.
- The average unemployment rate during the COVID-19 period increased to 17.77%.
- This represents an increase of approximately 8.26 percentage points.
- There were significant differences in average unemployment rates across regions.
- The analysis also showed differences between rural and urban areas.
- Monthly analysis revealed fluctuations in unemployment rates over time.
- The correlation analysis helped examine the relationship between unemployment, employment, and labour participation.

## Conclusion

This project provided an analysis of unemployment trends in India using Python.

The data was cleaned by handling missing values and converting dates into a suitable format for time-series analysis. Different visualizations were used to understand unemployment trends, regional differences, rural and urban patterns, and the impact of COVID-19.

The analysis showed a clear increase in unemployment during the COVID-19 period compared with the period before the lockdown. Overall, this project helped demonstrate how data analysis and visualization can be used to identify important patterns and derive meaningful insights from real-world economic data.
